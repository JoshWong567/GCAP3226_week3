## Task 1 — Stacked bar (not the demo pair)

In the demo, the stacked bar used **`government_consideration`** (groups) × **`support_info`** (stack, row %).

```python
# Task 1 — Stacked bar: support_info within each level of fairness
# Keep only complete cases for the grouping and stacked variables.
plot_data = df[["fairness", "support_info"]].dropna().copy()

# Count responses in each fairness × support_info combination, then convert
# each fairness row to percentages so every bar has a height of 100%.
percentage_table = pd.crosstab(
    plot_data["fairness"],
    plot_data["support_info"],
    normalize="index",
) * 100

# Count observations per fairness group for informative x-axis labels.
group_counts = plot_data.groupby("fairness").size()
percentage_table = percentage_table.sort_index()
percentage_table.index = [
    f"{level} (n={group_counts[level]})" for level in percentage_table.index
]

# Draw the row-percentage stacked bar chart.
ax = percentage_table.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5),
    colormap="viridis",
)
ax.set_title("Support for Information by Fairness Level")
ax.set_xlabel("Fairness level")
ax.set_ylabel("Percentage within fairness group (%)")
ax.set_ylim(0, 100)
ax.legend(title="Support for information", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Task 2 — Multiple linear regression (+ policy & severity)

In the demo you fitted multiple linear regression with `government_consideration` and `fairness` (and a fuller model with demographics).

**Your job:** fit a multiple linear regression for this sample:

**Response:** `support_info`  

**Exploratory variables (at least these four):**
- `government_consideration`
- `fairness`
- `policy_helpfulness`
- `waste_severity`

Use complete cases for these columns. Print **each coefficient**, **R²**, and **n**.  

In the **next code cell**, use the comment prompt with Inline Chat (see **How to use AI** above).


```python
# Task 2 — Multiple linear regression
# --- Sample AI prompt (Input → Process → Output) ---
# Input: dataframe df already loaded from week3.csv.
# Process: Write Python code to run a multiple linear regression with
#   response = support_info and exploratory variables =
#   government_consideration, fairness, policy_helpfulness, waste_severity.
#   Use rows with no missing values on these columns.
#   Print coefficients, R-squared, and n.
#   Add comments to key Python code.
# Output: A clear coefficient table (or printed list), R², and sample size n.
# --- End prompt ---

# Keep only complete observations for the response and four predictors.
regression_columns = [
    "support_info",
    "government_consideration",
    "fairness",
    "policy_helpfulness",
    "waste_severity",
]
regression_data = df[regression_columns].dropna()

# Separate the response from the explanatory variables and fit the model.
response = regression_data["support_info"]
predictors = regression_data[
    [
        "government_consideration",
        "fairness",
        "policy_helpfulness",
        "waste_severity",
    ]
]
model = LinearRegression()
model.fit(predictors, response)

# Present the fitted coefficients, model fit, and complete-case sample size.
coefficient_table = pd.DataFrame(
    {"coefficient": model.coef_},
    index=predictors.columns,
)
print("Intercept:", model.intercept_)
print("Coefficients:")
print(coefficient_table)
print("R-squared:", model.score(predictors, response))
print("n:", len(regression_data))
```
